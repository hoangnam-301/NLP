```python
import torch
import numpy as np

# 1. Tạo tensor từ list
data = [[1, 2], [3, 4]]
x_data = torch.tensor(data)
print(f"Tensor từ list:\n {x_data}\n")

# 2. Tạo tensor từ NumPy array
np_array = np.array(data)
x_np = torch.from_numpy(np_array)
print(f"Tensor từ NumPy array:\n {x_np}\n")

# 3. Tạo tensor với các giá trị ngẫu nhiên hoặc hằng số
x_ones = torch.ones_like(x_data) # tạo tensor toàn số 1, cùng shape với x_data
print(f"Ones Tensor:\n {x_ones}\n")

x_rand = torch.rand_like(x_data, dtype=torch.float) # tạo tensor ngẫu nhiên [0, 1)
print(f"Random Tensor:\n {x_rand}\n")

# 4. Kiểm tra thuộc tính
print(f"Shape của tensor: {x_rand.shape}")
print(f"Datatype của tensor: {x_rand.dtype}")
print(f"Device lưu trữ tensor: {x_rand.device}")
```

```python
# 1. Cộng x_data với chính nó
sum_tensor = x_data + x_data
print(f"Cộng x_data với chính nó:\n {sum_tensor}\n")

# 2. Nhân x_data với 5 (Scalar multiplication)
mul_scalar = x_data * 5
print(f"Nhân x_data với 5:\n {mul_scalar}\n")

# 3. Nhân ma trận x_data với x_data.T (Matrix Multiplication)
# x_data.T là ma trận chuyển vị
matrix_mul = x_data @ x_data.T
# Hoặc dùng: matrix_mul = torch.matmul(x_data, x_data.T)
print(f"Nhân ma trận x_data với chuyển vị của nó:\n {matrix_mul}")
```

```python
# Giả sử x_data là [[1, 2], [3, 4]]

# 1. Lấy ra hàng đầu tiên (index 0)
first_row = x_data[0]
print(f"Hàng đầu tiên: {first_row}")

# 2. Lấy ra cột thứ hai (tất cả các hàng, index cột là 1)
second_column = x_data[:, 1]
print(f"Cột thứ hai: {second_column}")

# 3. Lấy ra giá trị ở hàng thứ hai, cột thứ hai (index 1, 1)
last_value = x_data[1, 1]
print(f"Giá trị tại hàng 2, cột 2: {last_value}")
```

```python
import torch

# Tạo một tensor và yêu cầu tính đạo hàm cho nó (requires_grad=True)
x = torch.ones(1, requires_grad=True)
print(f"x: {x}")

# Thực hiện phép toán: y = x + 2
y = x + 2
print(f"y: {y}")
print(f"grad_fn của y: {y.grad_fn}") # Cho biết y được tạo từ phép cộng (Add)

# Thực hiện thêm phép toán: z = y * y * 3 => z = 3 * (x + 2)^2
z = y * y * 3
print(f"z: {z}")

# Tính đạo hàm của z theo x
z.backward()

# Kết quả đạo hàm dz/dx tại x=1
print(f"Đạo hàm của z theo x: {x.grad}")
```

```python
import torch
from torch import nn

# Khởi tạo lớp Linear: 5 đầu vào -> 2 đầu ra
linear_layer = nn.Linear(in_features=5, out_features=2)

# Tạo tensor đầu vào mẫu: 3 mẫu dữ liệu, mỗi mẫu có 5 đặc trưng
input_tensor = torch.randn(3, 5)

# Truyền qua lớp linear
output = linear_layer(input_tensor)

print(f"Input shape: {input_tensor.shape}") # torch.Size([3, 5])
print(f"Output shape: {output.shape}")      # torch.Size([3, 2])
print(f"Output:\n {output}")
```

```python
# 10 từ trong từ điển, mỗi từ là một vector 3 chiều
embedding_layer = nn.Embedding(num_embeddings=10, embedding_dim=3)

# Giả sử ta có một câu gồm 4 từ (chỉ số 1, 5, 0, 8)
input_indices = torch.LongTensor([1, 5, 0, 8])

# Lấy vector embedding
embeddings = embedding_layer(input_indices)

print(f"Input shape: {input_indices.shape}") # torch.Size([4])
print(f"Output shape: {embeddings.shape}")      # torch.Size([4, 3])
print(f"Embeddings:\n {embeddings}")
```

```python
class MyFirstModel(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim, output_dim):
        super(MyFirstModel, self).__init__()
        # Khai báo các thành phần
        self.embedding = nn.Embedding(vocab_size, embedding_dim)
        self.linear = nn.Linear(embedding_dim, hidden_dim)
        self.activation = nn.ReLU() # Hàm kích hoạt giúp mô hình học được các hàm phi tuyến
        self.output_layer = nn.Linear(hidden_dim, output_dim)

    def forward(self, indices):
        # 1. Chuyển ID từ thành Vector
        embeds = self.embedding(indices)
        # 2. Đi qua lớp ẩn và hàm kích hoạt
        hidden = self.activation(self.linear(embeds))
        # 3. Đi qua lớp đầu ra
        output = self.output_layer(hidden)
        return output

# Khởi tạo mô hình
model = MyFirstModel(vocab_size=100, embedding_dim=16, hidden_dim=8, output_dim=2)

# Kiểm tra với dữ liệu mẫu (1 câu có 4 từ)
input_data = torch.LongTensor([[1, 2, 5, 9]])
output_data = model(input_data)

print(f"Model output shape: {output_data.shape}") # torch.Size([1, 4, 2])
print(f"Kết quả dự báo đầu ra:\n {output_data}")
```
